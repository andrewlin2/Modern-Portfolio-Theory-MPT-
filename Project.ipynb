# Section 1: Motivation and Background

This project applies **Modern Portfolio Theory (MPT)** to construct and optimize a stock portfolio using real market data. The central question it addresses is: *given a set of stocks, how should an investor allocate capital across them to **maximize return for a given level of risk**, or equivalently, **minimize risk for a given level of return**?* 

The mathematical engine behind this is the **covariance matrix** — a structure that captures how every pair of stocks moves relative to one another. By analyzing these relationships, we can compute the **variance of any portfolio as a function of asset weights** and search over all possible allocations to identify those lying on the **Efficient Frontier** — the set of portfolios offering the best possible return at each level of risk. This allows us to pinpoint two key portfolios: the **Minimum Variance Portfolio** and the **Maximum Sharpe Ratio Portfolio**.

# Section 3 - Modern Portfolio Theory (MPT)

Modern Portfolio Theory (MPT) seeks to find the optimal combination of stocks to created a diversifed portfolio. MPT uses variance as a measure of risk for a stock: The higher the variance, the more volatile the stock is (the more the price fluctuates). It does this by calculating the covariance matrix between between stocks and looking for pairs (x, y) where COV(x, y) is close to -1 signaling that they behave oppositely.

## Section 3.1: Calculated Covariance Matrix

The covariance matrix is a matrix which relates behaviors of one stock with that of another stock. The covariance between two assets \(i\) and \(j\) is:

$$
\text{cov}(R_i, R_j) = \frac{1}{T-1} \sum_{t=1}^{T} (R_{i,t} - \bar{R}_i)(R_{j,t} - \bar{R}_j)
$$

The covariance matrix for \(n\) assets is:

$$
\Sigma =
\begin{bmatrix}
\text{var}(R_1) & \text{cov}(R_1, R_2) & \dots & \text{cov}(R_1, R_n) \\
\text{cov}(R_2, R_1) & \text{var}(R_2) & \dots & \text{cov}(R_2, R_n) \\
\vdots & \vdots & \ddots & \vdots \\
\text{cov}(R_n, R_1) & \text{cov}(R_n, R_2) & \dots & \text{var}(R_n)
\end{bmatrix}
$$

Where:

- \(R_{i,t}\) = return of asset \(i\) at time \(t\)  
- \(\bar{R}_i\) = mean return of asset \(i\)  
- \(T\) = total number of periods (trading days)

For this test, the testing period is days within the fiscal year 2023 for which yfinance was able to pull close price data. We have already converted the value to a percentage return, reflective of the precentage increase (positive %) or decrease (negative %) from the previous day's closing price. The pandas library already has a .cov() function that we can call. However, we must first remove the "Date" column from the dataframe.


```python
# calculate the average returns for each stock over a year

# drop the "Date" column
returns_2023.drop(columns=['Date'], inplace=True) 
print(returns_2023.head())

# calculate the matrix
covMatrix_2023 = returns_2023.cov()
print(covMatrix_2023.head())

def annualized_returns(df, trading_days=252):
    """
    Calculate annualized returns for each stock from daily % returns.
    
    Assumes df values are decimal returns (e.g. 0.01 = 1%).
    """
    cumulative_returns = (1 + df).prod()
    n_days = len(df)
    annualized = cumulative_returns ** (trading_days / n_days) - 1
    return annualized
    
annualized = annualized_returns(returns_2023)
print(annualized)
```

## Section 3.2: Application of Modern Portfolio Theory (MPT) Mathematical Approach

```python
"""
Risk-Minimizing Portfolio for a Target Return
==============================================
Workflow:
  1. Load your full covariance matrix (all stocks)
  2. Pick any subset of tickers
  3. Call min_risk_target_return() — it slices the sub-matrix automatically
"""

import numpy as np
import pandas as pd
from scipy.optimize import minimize

ourstocks = ["AMZN", "NVDA", "META", "AAPL", "TSLA"]
targetreturn = 0.7
allowshort = False

# example function call
# min_risk_target_return(covMatrix_2023,annualized,ourstocks,targetreturn,allowshort)

def min_risk_target_return(
    full_cov_matrix: pd.DataFrame,
    expected_returns: pd.Series,
    selected_tickers: list,
    target_return: float,
    allow_short: bool = False,
) -> dict:
    """
    Slice the covariance matrix to only the selected tickers, then find
    the minimum-risk portfolio that achieves the target return.

    Parameters
    ----------
    full_cov_matrix  : pd.DataFrame — your complete N×N covariance matrix
    expected_returns : pd.Series    — annualised mean return per ticker (all stocks)
    selected_tickers : list         — tickers you want in this portfolio
    target_return    : float        — desired annualised return (e.g. 0.10 for 10%)
    allow_short      : bool         — allow negative weights (default False)

    Returns
    -------
    dict with keys:
        "weights"          pd.Series  — optimal weight per selected ticker
        "expected_return"  float
        "volatility"       float
        "variance"         float
        "cov_matrix_used"  pd.DataFrame — the sub-matrix actually used
    """

    # ── 1. Validate selected tickers exist in the full matrix ─────────────
    missing_cov = [t for t in selected_tickers if t not in full_cov_matrix.columns]
    missing_ret = [t for t in selected_tickers if t not in expected_returns.index]

    if missing_cov:
        raise ValueError(f"These tickers are not in the covariance matrix: {missing_cov}")
    if missing_ret:
        raise ValueError(f"These tickers have no expected return: {missing_ret}")

    # ── 2. Slice the sub-covariance matrix ────────────────────────────────
    #       This is the core fix — extract only the rows AND columns for the
    #       selected tickers, preserving their covariance relationships.
    cov_matrix = full_cov_matrix.loc[selected_tickers, selected_tickers]

    # ── 3. Validate the sliced matrix ─────────────────────────────────────
    if cov_matrix.isnull().any().any():
        bad = cov_matrix.columns[cov_matrix.isnull().any()].tolist()
        raise ValueError(f"Sub-matrix has NaN values for tickers: {bad}")

    zero_diag = [t for t in selected_tickers if cov_matrix.loc[t, t] == 0]
    if zero_diag:
        raise ValueError(f"Zero variance (diagonal) for tickers: {zero_diag}")

    # ── 4. Align expected returns to selected tickers ─────────────────────
    mu    = expected_returns.reindex(selected_tickers).values.astype(float)
    Sigma = cov_matrix.values.astype(float)
    n     = len(selected_tickers)

    # ── 5. Check target return is achievable ──────────────────────────────
    mu_min = _min_var_return(Sigma, mu, n, allow_short)
    mu_max = float(mu.max())

    if not (mu_min - 1e-8 <= target_return <= mu_max + 1e-8):
        raise ValueError(
            f"\ntarget_return={target_return:.4%} is outside the achievable range "
            f"for {selected_tickers}.\n"
            f"  Min (global min-variance portfolio) : {mu_min:.4%}\n"
            f"  Max (best single asset)             : {mu_max:.4%}"
        )

    # ── 6. Optimise ───────────────────────────────────────────────────────
    constraints = [
        {"type": "eq", "fun": lambda w: float(w.sum() - 1.0)},
        {"type": "eq", "fun": lambda w: float(w @ mu - target_return)},
    ]
    bounds = [(-1.0, 1.0) if allow_short else (0.0, 1.0)] * n

    result = minimize(
        fun=lambda w: float(w @ Sigma @ w),
        x0=np.ones(n) / n,
        method="SLSQP",
        bounds=bounds,
        constraints=constraints,
        options={"ftol": 1e-12, "maxiter": 1000},
    )

    if not result.success:
        raise RuntimeError(f"Optimiser failed: {result.message}")

    # ── 7. Package results ────────────────────────────────────────────────
    w_opt = result.x.copy()
    w_opt[np.abs(w_opt) < 1e-6] = 0.0

    variance   = float(w_opt @ Sigma @ w_opt)
    volatility = float(np.sqrt(variance))

    return {
        "weights":         pd.Series(w_opt, index=selected_tickers, name="weights"),
        "expected_return": float(w_opt @ mu),
        "volatility":      volatility,
        "variance":        variance,
        "cov_matrix_used": cov_matrix,   # inspect this to confirm correct slicing
    }


def _min_var_return(Sigma, mu, n, allow_short):
    res = minimize(
        fun=lambda w: float(w @ Sigma @ w),
        x0=np.ones(n) / n,
        method="SLSQP",
        bounds=[(-1.0, 1.0) if allow_short else (0.0, 1.0)] * n,
        constraints=[{"type": "eq", "fun": lambda w: float(w.sum() - 1.0)}],
        options={"ftol": 1e-12, "maxiter": 1000},
    )
    return float(res.x @ mu)


min_risk_target_return(covMatrix_2023,annualized,ourstocks,targetreturn,allowshort) 

# ── Demo ──────────────────────────────────────────────────────────────────

# if __name__ == "__main__":

#     # Simulated full covariance matrix (replace with your real one)
#     all_tickers = ["AA", "AAL", "AAP", "AAPL", "ABBV", "AMZN", "HOOD", "META", "TSLA"]

#     np.random.seed(42)
#     rand = np.random.randn(len(all_tickers), len(all_tickers))
#     fake_cov = (rand @ rand.T) * 1e-4
#     full_cov_matrix = pd.DataFrame(fake_cov, index=all_tickers, columns=all_tickers)

#     all_returns = pd.Series({
#         "AA":   0.08, "AAL":  0.12, "AAP":  0.07, "AAPL": 0.15,
#         "ABBV": 0.06, "AMZN": 0.14, "HOOD": 0.18, "META": 0.20, "TSLA": 0.22,
#     })

#     # ── Choose any subset of stocks here ─────────────────────────────────
#     my_stocks = ["AMZN", "HOOD", "META", "AAPL", "TSLA"]

#     result = min_risk_target_return(
#         full_cov_matrix  = full_cov_matrix,
#         expected_returns = all_returns,
#         selected_tickers = my_stocks,
#         target_return    = 0.17,
#     )

#     print(f"\n{'─'*42}")
#     print(f"  Stocks selected  : {my_stocks}")
#     print(f"{'─'*42}")
#     print(f"  Achieved Return  : {result['expected_return']:.4%}")
#     print(f"  Volatility (σ)   : {result['volatility']:.4%}")
#     print(f"  Variance   (σ²)  : {result['variance']:.6f}")
#     print(f"{'─'*42}")
#     print("  Weights:")
#     for ticker, w in result["weights"].items():
#         bar = "█" * int(w * 40)
#         print(f"    {ticker:<6} {w:>7.2%}  {bar}")
#     print(f"{'─'*42}")
#     print(f"  Weights sum to   : {result['weights'].sum():.6f}")

#     print(f"\n  Sub-covariance matrix used ({len(my_stocks)}×{len(my_stocks)}):")
#     print(result["cov_matrix_used"].round(6))
```
