```python
import pandas as pd 
company_info = pd.read_csv("fortune1000_2024.csv").head(500) #fortune 1000 companies, we need something to call (yfinance). 

print(company_info.shape)
company_info.head()
```

```python
import yfinance as yf  #yahoo finance, call functions that retrive data for me

# Prepare tickers
tickers = company_info['Ticker'].dropna().astype(str).tolist() # company_info['Ticker'] is going to return the column that has column name Ticker
                                # dropna drops anything that's blank
                                #
# Clean the list to all tickers work with yfinance
list_to_drop = ['NYCB', 'CLR', 'GPS', 'PXD', 'BECN', 'SQ', 'HES', 'DFS', 'RADCQ', 'SKX', 'QRTEA', 'X', 'WBA', 'LSXMA', 'JWN', 'BERY', 'BRKA', 'SPTN', 'PARA', 'CHK', 'FL','IPG', 'FI', 'AMRK', 'OMI', 'ATUS', 'K', 'ODP', 'MMC']
for ticker in tickers:
    if ticker in list_to_drop:
        tickers.remove(ticker)
#some tickers in yfinance doesn't work anymore (delisted companies, error in ticker, etc). Removes any of those

# Download closing prices
data_2023 = yf.download(tickers, start="2023-01-01", end="2023-12-31")['Close'] # does not have to call yfinance everytime
data_2024 = yf.download(tickers, start="2024-01-01", end="2024-12-31")['Close'] # we dont have opening price here because 

data_2023 = data_2023.dropna(axis=1, how='all') # when we get this data, there is extra information we want to drop. <date> <closing>
data_2024 = data_2024.dropna(axis=1, how='all')

returns_2023 = data_2023.pct_change(fill_method=None).dropna() # calculating returns for every stock 
returns_2024 = data_2024.pct_change(fill_method=None).dropna() # change between closing price and opening price on a trading day

print(returns_2023.head())
print(returns_2024.head())

returns_2023.to_csv("returns_2023.csv")
returns_2024.to_csv("returns_2024.csv")
```
